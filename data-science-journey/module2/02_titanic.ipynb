```python
import pandas as pd

URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)
df.head()
```

# Question
**1. Aperçu global : Combien de passagers ? Combien de NaN par colonne ?**

```python
df.info()
df.isna().sum()
```

**2.Taux de survie global**

```python
tx_survie = round(df["Survived"].mean()*100,2)
print("Le taux de survie global est " + str(tx_survie) + "%")
```

**3.Taux de survie par classe**

```python
df.groupby("Pclass")["Survived"].mean()
```

**4.Taux de survie par sexe**

```python
df.groupby("Sex")["Survived"].mean()
```

**5.Taux de survie par classe et sexe**

```python
df.groupby(["Pclass","Sex"])["Survived"].mean()
```

**6.Tranches d'age**

```python
def age_group(age):
    if pd.isna(age):
        return "inconnu"    
    elif age < 18:
        return "enfant"
    elif age < 60:
        return "adulte"
    else:
        return "senior"
df["Age_group"] = df["Age"].apply(age_group)
df.groupby("Age_group")["Survived"].mean()
```
