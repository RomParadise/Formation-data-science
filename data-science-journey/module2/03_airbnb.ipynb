```python
import pandas as pd

URL = "http://data.insideairbnb.com/france/ile-de-france/paris/2024-09-06/visualisations/listings.csv"
df = pd.read_csv(URL)
df.head()
```

```python
df.info()
```

**Filtres logiques : garde uniquement les logements où :**
- price entre 20 et 1000 (on enlève les valeurs aberrantes)
- minimum_nights <= 30

```python
df = df[(df["price"]> 20) & (df["price"] < 1000) & (df["minimum_nights"] <= 30)]
```

**Top 10 quartiers par prix médian :**

```python
df.groupby("neighbourhood")["price"].median().sort_values(ascending=False).head(10)
```

**Bonus : Prix médian par room_type (Entire home, Private room, etc.).**

```python
df.groupby("room_type")["price"].median().sort_values(ascending=False)
```
